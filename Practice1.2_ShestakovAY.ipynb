#### **Задание 1.** Создание двух массивов и проверка арифметических операций

```python
mas1 = np.array([34, 45, 56, 0, -1])
mas2 = np.array([10, -5, 3, 8, 20])

print("mas1 =", mas1)
print("mas2 =", mas2)

print("Сложение:      mas1 + mas2 =", mas1 + mas2)
print("Вычитание:     mas1 - mas2 =", mas1 - mas2)
print("Умножение:     mas1 * mas2 =", mas1 * mas2)
print("Деление:       mas1 / mas2 =", mas1 / mas2)
print("Целочисл. дел: mas1 // mas2 =", mas1 // mas2)
print("Остаток:       mas1 % mas2 =", mas1 % mas2)
print("Степень:       mas1 ** 2 =", mas1 ** 2)
```

#### **Задание 2.** Проверка операций для разных типов данных

```python
# Вещественные числа
mas_float1 = np.array([1.5, -2.3, 0.0, 4.8, 10.1])
mas_float2 = np.array([2.0, 3.3, 1.1, -0.5, 5.0])
print("Вещественные: сложение =", mas_float1 + mas_float2)
print("Вещественные: вычитание =", mas_float1 - mas_float2)
print("Вещественные: умножение =", mas_float1 * mas_float2)

# Комплексные числа
mas_complex1 = np.array([4+5j, -8+1j, 2-3j])
mas_complex2 = np.array([1+1j, 2+2j, -1+0j])
print("Комплексные: сложение =", mas_complex1 + mas_complex2)
print("Комплексные: умножение =", mas_complex1 * mas_complex2)

# Строковые данные — работает только сложение
mas_str1 = np.array(["Привет", "мир", "Python"])
mas_str2 = np.array([", ", " - ", "!"])
print("Строковые: сложение (конкатенация) =", np.char.add(mas_str1, mas_str2))

# Логические значения — работают как 0/1
mas_bool1 = np.array([True, False, True])
mas_bool2 = np.array([True, True, False])
print("Логические: сложение =", mas_bool1 + mas_bool2)
print("Логические: умножение =", mas_bool1 * mas_bool2)
```

#### **Задание 3.** Тригонометрические и другие функции Numpy

#### **Задание 5.** Индексация массива, элементы в обратном порядке, сумма положительных

```python
A = np.array([5, -3, 8, -1, 0, 12, -7, 4])
print("Массив A =", A)

# Элементы массива в обратном порядке
print("A[-1], A[-2], ... :", [A[-i] for i in range(1, len(A) + 1)])

# Сумма положительных элементов
positive_sum = A[A > 0].sum()
print("Сумма положительных элементов A =", positive_sum)
```
